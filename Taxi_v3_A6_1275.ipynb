Create the environment

```python
import gymnasium as gym
env = gym.make("Taxi-v4")
print("Environment Created")
```

Reset the Environment

```python
state, info = env.reset()
print(state)
taxi_row, taxi_col, pickup, dest = env.unwrapped.decode(state)
print(f"taxi_row={taxi_row}")
print(f"taxi_col={taxi_col}")
print(f"pickup={pickup}")
print(f"dest={dest}")
```

Understanding the Action Space

```python
actions = env.action_space
print(actions)
```

Random Action


```python
action = env.action_space.sample()
print(action)

if action == 0:
    print("move South")

if action == 1:
    print("move North")

if action == 2:
    print("move East")

if action == 3:
    print("move West")

if action == 4:
    print("move Pickup")

if action == 5:
    print("move Drop")
```

Agent Act

```python
next_state, reward, terminated, truncated, info = env.step(action)
print(next_state)
print(reward)
print(state)
```

Random Agent

```python
import gymnasium as gym

env = gym.make("Taxi-v4")

state, info = env.reset()
current_state = state

episodes = 20
total_reward = 0

for episode in range(episodes):
    action = env.action_space.sample()
    next_state, reward, terminated, truncated, info = env.step(action)

    total_reward += reward

    if terminated or truncated:
        break

    current_state = next_state

print(f"Total reward={total_reward}")
print(f"Average reward={total_reward/(episode+1)}")
```
