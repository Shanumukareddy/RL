```python

# Value Iteration in a Deterministic 2x2 Grid World

# 1. Define the MDP
states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]
goal = "G"
gamma = 0.9

# Transition rules
transitions = {
    "S0": {
        "Up": "S0",       # Wall
        "Down": "S1",
        "Left": "S0",     # Boundary
        "Right": "S0"     # Wall
    },
    "S1": {
        "Up": "S0",
        "Down": "S1",     # Boundary
        "Left": "S1",     # Boundary
        "Right": "G"
    }
}

# Reward function
def get_reward(next_state):
    if next_state == "G":
        return 100
    return -1

# 2. Initialize state values
V = {"S0": 0, "S1": 0, "G": 0}

# 3. Value Iteration
theta = 0.001
iteration = 0

print("Iteration-wise State Values")
print("-" * 45)
print(f"{'Iteration':<12}{'V(S0)':<15}{'V(S1)':<15}")

while True:
    iteration += 1
    new_V = V.copy()
    delta = 0

    for state in states:
        action_values = []

        for action in actions:
            next_state = transitions[state][action]
            reward = get_reward(next_state)

            q_value = reward + gamma * V[next_state]
            action_values.append(q_value)

        new_V[state] = max(action_values)

        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V

    print(f"{iteration:<12}{V['S0']:<15.2f}{V['S1']:<15.2f}")

    if delta < theta:
        break

# 4. Calculate final Q-values
Q = {}

for state in states:
    Q[state] = {}

    for action in actions:
        next_state = transitions[state][action]
        reward = get_reward(next_state)

        Q[state][action] = reward + gamma * V[next_state]

# 5. Find optimal policy
policy = {}

for state in states:
    best_value = max(Q[state].values())

    best_actions = [
        action for action in actions
        if abs(Q[state][action] - best_value) < 1e-9
    ]

    policy[state] = best_actions

# 6. Display results

print("\nFinal Optimal State Values")
for state in states:
    print(f"V*({state}) = {V[state]:.2f}")

print(f"V*(G) = {V['G']:.2f}")

print("\nFinal Action-Value Table Q*(s,a)")
for state in states:
    print(f"\nState: {state}")

    for action in actions:
        print(f"Q({state}, {action}) = {Q[state][action]:.2f}")

print("\nOptimal Policy")
for state in states:
    print(f"{state}: {', '.join(policy[state])}")

print("\nOptimal Path: S0 -> S1 -> G")
```
